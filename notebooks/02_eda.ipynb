# CreditWise — 02. Exploratory Data Analysis

Class balance, categorical breakdowns, distributions, and the correlation heatmap. Uses the cleaned/encoded data produced by `01_data_preprocessing.ipynb`.

```python
import sys
sys.path.append("..")

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from src.preprocessing import load_data
```

```python
df_raw = load_data("../data/raw/loan_approval_data.csv")
df_encoded = pd.read_csv("../data/processed/cleaned_encoded.csv")
```

## Class balance

How many applications were approved vs. rejected.

```python
classes_count = df_raw["Loan_Approved"].value_counts()
plt.pie(classes_count, labels=classes_count.index, autopct="%1.1f%%")
plt.title("Loan Approved?")
plt.show()
```

## Categorical breakdown — Gender

```python
ax = sns.barplot(x=df_raw["Gender"].value_counts().index, y=df_raw["Gender"].value_counts().values)
ax.bar_label(ax.containers[0])
ax.set_ylabel("count")
plt.title("Applicants by Gender")
plt.show()
```

## Income distribution

```python
sns.histplot(data=df_raw, x="Applicant_Income", bins=20)
plt.title("Applicant Income Distribution")
plt.show()
```

## Loan amount by approval outcome

```python
sns.boxplot(data=df_raw, x="Loan_Approved", y="Loan_Amount")
plt.title("Loan Amount by Approval Outcome")
plt.show()
```

## Correlation heatmap

**Bug fix:** the original notebook computed `corr_matrix` but then printed `corr_metrix` (typo) in the next cell, which raised a `NameError`. Fixed here.

```python
num_cols = df_encoded.select_dtypes(include="number")
corr_matrix = num_cols.corr()

plt.figure(figsize=(15, 8))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Matrix (encoded features)")
plt.show()
```
