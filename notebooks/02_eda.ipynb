# CreditWise — 02. Exploratory Data Analysis

Class balance, categorical breakdowns, distributions, and the correlation heatmap. Uses the cleaned/encoded data produced by `01_data_preprocessing.ipynb`.

```python
import sys
sys.path.append("..")

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from src.preprocessing import load_data
```

```python
df_raw = load_data("../data/raw/loan_approval_data.csv")
df_encoded = pd.read_csv("../data/processed/cleaned_encoded.csv")
```

## Class balance

How many applications were approved vs. rejected.

```python
classes_count = df_raw["Loan_Approved"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    classes_count,
    labels=classes_count.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Loan Approved?")
plt.show()
```

## Categorical breakdown — Gender

```python
gender_counts = df_raw["Gender"].value_counts()

ax = sns.barplot(
    x=gender_counts.index,
    y=gender_counts.values
)

ax.bar_label(ax.containers[0])
ax.set_ylabel("count")
ax.set_xlabel("Gender")
plt.title("Applicants by Gender")
plt.show()
```

## Income distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df_raw, x="Applicant_Income", bins=20, kde=True)
plt.title("Applicant Income Distribution")
plt.xlabel("Applicant Income")
plt.ylabel("Count")
plt.show()
```

## Loan amount by approval outcome

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_raw, x="Loan_Approved", y="Loan_Amount")
plt.title("Loan Amount by Approval Outcome")
plt.xlabel("Loan Approved")
plt.ylabel("Loan Amount")
plt.show()
```

## Correlation heatmap

Correlation between the encoded numerical features and loan approval.

```python
num_cols = df_encoded.select_dtypes(include="number")
corr_matrix = num_cols.corr()

plt.figure(figsize=(15, 8))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title("Correlation Matrix (Encoded Features)")
plt.tight_layout()
plt.show()
```
