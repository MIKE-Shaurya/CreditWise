# CreditWise — 04. Modeling & Evaluation

Trains Logistic Regression, KNN, and Naive Bayes on both the baseline (cleaned/encoded) data and the engineered data, and compares them side by side.

**Bug fix:** the original notebook's feature-engineering comparison only re-trained KNN, so there was no way to tell whether the engineered features helped or hurt the other two models. All three are compared here, on both feature sets.

```python
import sys
sys.path.append("..")

import pandas as pd

from src.modeling import split_and_scale, train_and_evaluate_all
```

## Baseline model (cleaned/encoded features only)

```python
df_base = pd.read_csv("../data/processed/cleaned_encoded.csv")

X_train, X_test, y_train, y_test, _ = split_and_scale(df_base)
baseline_results = train_and_evaluate_all(X_train, X_test, y_train, y_test)

for r in baseline_results:
    print(r)
    print()
```

## Engineered features

```python
df_fe = pd.read_csv("../data/processed/engineered.csv")

X_train_fe, X_test_fe, y_train_fe, y_test_fe, _ = split_and_scale(df_fe)
fe_results = train_and_evaluate_all(X_train_fe, X_test_fe, y_train_fe, y_test_fe)

for r in fe_results:
    print(r)
    print()
```

## Side-by-side comparison

```python
comparison = pd.DataFrame({
    "Model": [r.model_name for r in baseline_results],
    "Baseline Accuracy": [r.accuracy for r in baseline_results],
    "Engineered Accuracy": [r.accuracy for r in fe_results],
    "Baseline F1": [r.f1 for r in baseline_results],
    "Engineered F1": [r.f1 for r in fe_results],
})
comparison["Accuracy Delta"] = comparison["Engineered Accuracy"] - comparison["Baseline Accuracy"]
comparison["F1 Delta"] = comparison["Engineered F1"] - comparison["Baseline F1"]
comparison
```

Positive deltas mean the engineered features helped that model; negative means the baseline was better for that model. Unlike the original notebook — where the one engineered-feature test happened to make KNN *worse* because it squared `Credit_Score` in place — this table makes it easy to see the effect per model instead of assuming it generalizes.
